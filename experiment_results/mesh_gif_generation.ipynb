```python
import glob, os

base = ".."  

def list_renderable(dataset):
    pattern = f"{base}/outputs/{dataset}/sapien/*/final/train/ours_20000/meshes/part_0.ply"
    scenes = []
    for p in sorted(glob.glob(pattern)):
        scene = p.split("/sapien/")[1].split("/")[0]
        scenes.append(scene)
    return scenes

VideoArtGS_scenes = list_renderable("videoartgs")
V2a_scenes = list_renderable("v2a")

print("VideoArtGS scene number:", len(VideoArtGS_scenes))
print(VideoArtGS_scenes)
print("V2a scene number:", len(V2a_scenes))
print(V2a_scenes)
```

```python
for scene in VideoArtGS_scenes:
    generate_video_gif('videoartgs', 'sapien', scene)
    color_parts(scene, dataset="videoartgs", base="..")

for scene in V2a_scenes:
    generate_video_gif('v2a', 'sapien', scene)
    color_parts(scene, dataset="v2a", base="..")
```
