```python
import glob, os

base = ".."  
output_dir="outputs_PAT"
subset="sapien"
def list_renderable(dataset, output_dir, subset):
    pattern = f"{base}/{output_dir}/{dataset}/{subset}/*/final/train/ours_20000/meshes/part_0.ply"
    scenes = []
    for p in sorted(glob.glob(pattern)):
        scene = p.split(f"/{subset}/")[1].split("/")[0]
        scenes.append(scene)
    return scenes

VideoArtGS_scenes = list_renderable("videoartgs", output_dir, subset)
# V2a_scenes = list_renderable("v2a", "sapien")

print("VideoArtGS scene number with sapien:", len(VideoArtGS_scenes))
print(VideoArtGS_scenes)
# print("V2a scene number:", len(V2a_scenes))
# print(V2a_scenes)
```

```python
# VideoArtGS_scenes=["168"]
subset="sapien"
output_dir="outputs_PAT"
for scene in VideoArtGS_scenes:
    generate_video_gif(dataset='videoartgs', subset=subset, scene=scene, output_dir=output_dir)
    color_parts(scene, dataset="videoartgs", output_dir=output_dir, subset=subset, base="..")

# for scene in V2a_scenes:
#     generate_video_gif('v2a', 'sapien', scene)
#     color_parts(scene, dataset="v2a", output_dir=output_dir, subset=subset, base="..")
```
