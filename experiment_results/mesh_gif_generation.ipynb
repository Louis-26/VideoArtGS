```python
import glob, os

base = ".."  
subset="realscan"
def list_renderable(dataset, subset):
    pattern = f"{base}/outputs/{dataset}/{subset}/*/final/train/ours_20000/meshes/part_0.ply"
    scenes = []
    for p in sorted(glob.glob(pattern)):
        scene = p.split(f"/{subset}/")[1].split("/")[0]
        scenes.append(scene)
    return scenes

VideoArtGS_scenes = list_renderable("videoartgs", "realscan")
# V2a_scenes = list_renderable("v2a", "sapien")

print("VideoArtGS scene number with realscan:", len(VideoArtGS_scenes))
print(VideoArtGS_scenes)
# print("V2a scene number:", len(V2a_scenes))
# print(V2a_scenes)
```

```python
for scene in VideoArtGS_scenes:
    generate_video_gif('videoartgs', 'realscan', scene)
    color_parts(scene, dataset="videoartgs", subset="realscan", base="..")

# for scene in V2a_scenes:
#     generate_video_gif('v2a', 'sapien', scene)
#     color_parts(scene, dataset="v2a", base="..")
```
